## 02 Models & Messages
In this workbook we'll cover how Langchain supports various models & the various types of messages - viz `HumanMessage`, `SyetemMessage`, `AIMessage`.

```python
from dotenv import load_dotenv

# for some colorful text
from rich.console import Console
from rich.markdown import Markdown

from langchain.chat_models import init_chat_model

# load API keys from local .env file
load_dotenv(override=True)
console = Console()

# assuming you have the required API keys listed in your local .env
# file, here is how you can instantiate a chat model
openai_llm = init_chat_model(
    "gpt-4o-mini",  # replace this with any supported OpenAI model name
    model_provider="openai",
    temperature=0.0,
)

anthropic_llm = init_chat_model(
    "claude-haiku-4-5",  # replace this with any supported Anthropic model name
    model_provider="anthropic",
    temperature=0.0,
)

gemini_llm = init_chat_model(
    "google_genai:gemini-2.0-flash",  # replace this with any supported Google Gemini model name
    temperature=0.0,
)
```

```python
from langchain.agents import create_agent
from langchain_core.messages import HumanMessage, SystemMessage

agent = create_agent(
    model=openai_llm,  # you can switch this to anthropic_llm or gemini_llm as needed
    system_prompt=SystemMessage(
        content="""
        You are a very experienced local Mumbai guide, who knows all the hidden gems, 
        local foodspots & drinking holes of Mumbai. You respond in the local "tapori" 
        lingo, using words such as "aapun", "bindaas", "mast", "jhakaas", "dil se", 
        "full paisa vasool", "bole to" etc. Don't limit yourself to just these words, but feel 
        free to use any other "tapori" style words and phrases as you see fit - no 'gaalis' please!
        All responses must be in English ONLY but with a "tapori" flavor.
        """
    ),
)
```

```python
# a utility function to call the agent & return response
def call_agent(agent, query: str) -> str:
    human_message = HumanMessage(content=query)
    response = agent.invoke({"messages": [human_message]})
    # return content of last message returned
    return response  # ["messages"][-1].content
```

```python
# now let's call our agent & see how it responds to a user query about Mumbai food spots
response = call_agent(
    agent,
    query="Hey, how are you? I'm visiting Mumbai for the first time. Can you suggest 5 local food spots to check out?",
)
console.print(Markdown(response["messages"][-1].content))
```

```python
for msg in response["messages"]:
    console.print(f"[bright_yellow]{msg.type}:[/bright_yellow] {msg.content}")
```

### Alternative Formats
You need not always be explicit in declaring the type of message to use. LangChain is able to decipher the type of message from the context, as seen below.

```python
agent2 = create_agent(
    model=openai_llm,
    # This is a SystemMessage - need not wrap in SystemMessage("....")
    system_prompt=(
        """
        You are a sports expert who can provide information, detailed analysis and insights
        on all the popular sports played around the world, including football, basketball, cricket, tennis, and more.
        """
    ),
)
```

```python
response = agent2.invoke(
    {"messages": "Briefly describe the game of Cricket to a newbie"}
)
# Again, no need to specify HumanMessage("Briefly describe....") here
console.print(Markdown(response["messages"][-1].content))
```

```python
# you can use the dictionary style of passing messages
response = agent2.invoke(
    {
        "messages": {
            "role": "user",
            "content": "Which is the best team in the IPL?",
        }
    }
)
console.print(Markdown(response["messages"][-1].content))
```

```python
console.print(response["messages"][-1])  # let's see how a message looks like

# NOTE: console.print(...) call automatically formats the output message, so we can clearly
# see all the field & contents
```
