## 01 - Self Querying Retrieval Technique

In this technique, we have a step between the retrieval and the input. So the person types in their query and we then use an LLM to reformat that query to get both the semantic elements of the query, but also to be able to convert it so that we can do searches on metadata as we go through it.
This is a fundamental fact, that if you are looking for a movie and you want to specify the year, you don't want to use for year using a vector store & semantic serrch. Yoy want to do a lookup tht looks for year an filters the results back based on the year you put in.
Just the same as searches for music on Spotify. For example, if user gives you the name of the artist, you don't want to use semantic search to search tp lookup ame of artist - you want to do a query that looks for the artist and then use semantic search and then use semantic search where it is strong.
