# 01 First Steps with LangChain
In this notebook, we introduce first steps with LangChain such as creating a chat model, using `init_chat_model()` API call, `SystemMessage` and `HumanMessage`, `ChatPromptTemplate`s, enabling _thinking_ capabilities of more recent LLMs such as Claude Haiku 4.5 Open AI 03-mini and Gemini 3.5 Flash for example.

````python
from dotenv import load_dotenv
from rich.console import Console
from rich.markdown import Markdown

load_dotenv(override=True)
console = Console()
````

As a first example, we'll instantiate a Claude Haiku 4.5 chat model and ask it to generate code for a Python factorial function.

````python
from langchain.chat_models import init_chat_model
from langchain_core.messages import HumanMessage, SystemMessage
````

````python
chat_model = init_chat_model(
    "anthropic:claude-haiku-4-5",
    # model_provider="anthropic",
    temperature=0.0,
)

human_message = """
    Write me a Python program to calculate factorial of a number. 
    Handle all edge cases correctly!
    Return just the factorial function code and nothing else - not test harness, no main() function.
    Return code as markdown delimited with ```pyhon ... ``` for better readability.
"""

messages = [
    SystemMessage(content="You are a helpful programming assistant"),
    HumanMessage(content=human_message),
]

response = chat_model.invoke(messages)
console.print(f"[bright_blue]AI response: [/bright_blue]")
console.print(Markdown(response.content))
````

## Reasoning models
Anthropic’s Claude 3.7 Sonnet (or Haiku) introduced a powerful capability called extended thinking that allows the model to show its reasoning process before delivering a final answer.

This feature represents a significant advancement in how developers can leverage LLMs for complex reasoning tasks. Similar capabilities are offered by Open AI 03-mini and Gemini 3.5-flash models.

Here’s how to configure extended thinking through the ChatAnthropic

````python
from langchain_core.prompts import ChatPromptTemplate

# instantiate the model - we'll use Claude Haiku
reasoning_model = init_chat_model(
    "anthropic:claude-haiku-4-5",
    max_tokens=64_000,
    # reasoning tokens are now called "thinking" tokens in the latest LangChain versions
    # they are expensive, so use them wisely - they allow the model to "think" for a longer
    # time and produce more complex responses
    thinking={"type": "enabled", "budget_tokens": 15_000},
)

# --------------------------------------------------------
# to use OpenAI LLM instead, uncomment the following code
# ---------------------------------------------------------
# # Instantiate the OpenAI reasoning model
# reasoning_model = init_chat_model(
#     "openai:o3-mini",
#     # Alternatively you can use "openai:o1"
#     # Controls the amount of reasoning tokens allocated
#     reasoning_effort="medium",
# )

# ----------------------------------------------------------
# to use Gemini model instead, uncomment the following code
# ----------------------------------------------------------
# # Instantiate the Google Gemini reasoning model
# reasoning_model = init_chat_model(
#     "google_genai:gemini-3.5-flash",
#     # Adjust depth: "low", "medium", or "high"
#     thinking_level="high",
#     # Required for Gemini 3+ models when thinking is active
#     # to avoid execution bugs
#     temperature=1.0,
# )

prompt_template = ChatPromptTemplate.from_messages(
    [
        ("system", "You are a helpful assistant for solving coding problems."),
        ("human", "{human_message}"),
    ]
)

human_message = """
    Design an algorithm to find the kth largest element in an unsorted array
    with the optimal time complexity. Analyze the time and space complexity
    of your solution and explain why it's optimal.
    Then show me Python code for the algorithm, with adequate doc and inline comments.
    Also add a main() function to test the function. Code should be bracketed
    within ```python ... ``` delimiters for better readability.
    Return your response as properly formatted markdown.
"""

chain = prompt_template | reasoning_model

response = chain.invoke({"human_message": human_message})
````

````python
console.print(f"[yellow]AI response: [/yellow]")
console.print(response)
````

````python
console.print(Markdown(response.content[1]["text"]))
````
