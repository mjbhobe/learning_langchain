# Indexing
In previous notebooks, we covered the topic of [Query Translation](02+QueryTranslation.ipynb), where the goal was to take the input user question and to translate it in some way so as to improve retrival. We saw variuous techniques like Multi-Query, RAG Fusion, Decomposition, Step-back and HYDE. We also covered the topic of [Routing](03_Routing.ipynb), which is the process of taking the (translated) query and routing it to the correct source of the context - for example, directing Health Insurance queries to Health Insurance database, Car Insurance queries to the Car Insurance database and so on. Then we covered the topic of [Query Construction](04_Query_Construction.ipynb), we we convert a text query to a specific format (SQL, Cipher, vector etc.) depending on source of the context (an RDBMS, a GraphDB or a vector store respectively).

In this notebook, we'll cover topic of indexing, specifically the following techniques:
* Multi-representation
* RAPTOR
* ColBERT
